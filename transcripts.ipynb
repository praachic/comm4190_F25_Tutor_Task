# Showing chat transcripts

* There are three chat transcripts:
  1. No specific tutoring prompt
  2. Using Mollick Tutor Prompt
  3. Revised tutor prompt

```python
import json
import os
from IPython.display import Markdown, display
```

```python
chats = [fname for fname in os.listdir('.') if fname.endswith('.json')]
chats
```

```python
def display_chat(chat_json):
    title = chat_json[0]['chat']['title']
    display(Markdown(f'### {title}'))

    for message in chat_json[0]['chat']['history']['messages'].values():
        display(Markdown(f'#### {message['role'].title()}'))
        display(Markdown(message['content']))
        display(Markdown('---'))
```

## 1. No Prompt

```python
no_prompt = json.load(open('NoPromptTreeIdentifyRevised.json'))
```

```python
display_chat(no_prompt)
```

## 2. Mollick structured tutor prompt

```python
tutor_chat = json.load(open('TutorPromptTreeIdentify.json'))
```

```python
display_chat(tutor_chat)
```

## 3. Revised tutor prompt

```python
revised_chat = json.load(open('ReworkedPromptTreeIdentify.json'))
```

```python
display_chat(revised_chat)
```
